```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
df = spark.read.csv("/Volumes/nilesilverlayercat/default/nilebronzevolume/orderdetails.csv",header=True,inferSchema=True)


```

###  *Rename Columns*


```python
rename_columns = {
    "sale_row_id": "Sale_Row_ID",
    "invoice_id": "Invoice_ID",
    "transaction_date": "Date",
    "transaction_time": "Time",
    "branch": "Branch",
    "city": "City",
    "customer_type": "Customer_Type",
    "gender": "Gender",
    "product_category": "Product_Category",
    "product_name": "Product_Name",
    "unit_price": "Unit_Price",
    "quantity": "Quantity",
    "discount_percent": "Discount_Percent",
    "tax_rate": "Tax_Rate",
    "cost_per_unit": "Cost_Per_Unit",
    "payment_method": "Payment_Method",
    "customer_rating": "Customer_Rating"
}

for old_name, new_name in rename_columns.items():
    df = df.withColumnRenamed(old_name, new_name)

print(df.columns)
```

### *Missing Values Report*

```python
# --------------------------------------------
# Missing Values Report
# --------------------------------------------
missing_values_before = df.select([
    sum(
        when(
            col(column_name).isNull(),
            1
        ).otherwise(0)
    ).alias(column_name)
    for column_name in df.columns
])

display(missing_values_before)
```

```python
# --------------------------------------------
# Find String Columns
# --------------------------------------------
string_columns = [
    field.name
    for field in df.schema.fields
    if isinstance(field.dataType, StringType)
]

# --------------------------------------------
# Remove Leading and Trailing Spaces
# --------------------------------------------
for column_name in string_columns:
    df = df.withColumn(
        column_name,
        trim(col(column_name))
    )

# --------------------------------------------
# Replace Missing Text with Null
# --------------------------------------------
for column_name in string_columns:
    df = df.withColumn(
        column_name,
        when(
            col(column_name).isNull() |
            lower(col(column_name)).isin(
                "",
                "null",
                "none",
                "not available"
            ),
            None
        ).otherwise(col(column_name))
    )
```

```python
# --------------------------------------------
# Standardize Branch //|\\
# --------------------------------------------
branch_value = lower(
    regexp_replace(
        trim(col("Branch")),
        r"[\s-]+",
        ""
    )
)

df = df.withColumn(   #????///|\\\-------------------------------------------------------------------------------------------->
    "Branch",
    when(
        branch_value.isin("brancha", "a"),
        "Branch A"
    )
    .when(
        branch_value.isin("branchb", "b"),
        "Branch B"
    )
    .when(
        branch_value.isin("branchc", "c"),
        "Branch C"
    )
    .otherwise(None)
)
```

```python
# --------------------------------------------
# Standardize City
# --------------------------------------------
city_value = lower(trim(col("City")))

df = df.withColumn(
    "City",
    when(city_value == "cairo", "Cairo")
    .when(city_value == "giza", "Giza")
    .when(city_value == "alexandria", "Alexandria")
    .otherwise(None)
)
# to handle nulls
```

```python
# --------------------------------------------
# Fill Missing Branch Using City
# --------------------------------------------
df = df.withColumn(
    "Branch",
    when(
        col("Branch").isNull() &
        (col("City") == "Cairo"),
        "Branch A"
    )
    .when(
        col("Branch").isNull() &
        (col("City") == "Giza"),
        "Branch B"
    )
    .when(
        col("Branch").isNull() &
        (col("City") == "Alexandria"),
        "Branch C"
    )
    .otherwise(col("Branch"))
)

# --------------------------------------------
# Correct City Using Branch
# --------------------------------------------
df = df.withColumn(
    "City",
    when(col("Branch") == "Branch A", "Cairo")
    .when(col("Branch") == "Branch B", "Giza")
    .when(col("Branch") == "Branch C", "Alexandria")
    .otherwise(col("City")) #keep the same value and don't change it 
)
```

```python
# --------------------------------------------
# Standardize Customer Type
# --------------------------------------------
customer_value = lower(
    regexp_replace(
        trim(col("Customer_Type")),
        r"\s+",    # Replaces multiple spaces with one space.
        " "
    )
)

df = df.withColumn(
    "Customer_Type",
    when(
        customer_value.isin(
            "member",
            "loyal member"
        ),
        "Member"
    )
    .when(
        customer_value.isin(
            "normal",
            "regular",
            "non member"
        ),
        "Normal"
    )
    .otherwise("Unknown")
)
```

```python
# --------------------------------------------
# Standardize Gender
# --------------------------------------------
gender_value = lower(trim(col("Gender")))

df = df.withColumn(
    "Gender",
    when(
        gender_value.isin("male", "m"),
        "Male"
    )
    .when(
        gender_value.isin("female", "f"),
        "Female"
    )
    .otherwise("Unknown")
)
```

```python
# --------------------------------------------
# Standardize Product Category
# --------------------------------------------
category_value = lower(
    trim(
        regexp_replace(
            col("Product_Category"),
            "&",
            "and"
        )
    )
)

category_value = regexp_replace(
    category_value,
    r"\s+", # Replaces multiple spaces with one space.
    " "
)
# --------------------------------------------
#  
# --------------------------------------------
df = df.withColumn(
    "Product_Category",
    when(
        category_value == "food and beverages",
        "Food and Beverages"
    )
    .when(
        category_value == "health and beauty",
        "Health and Beauty"
    )
    .when(
        category_value == "electronics",
        "Electronics"
    )
    .when(
        category_value == "home and lifestyle",
        "Home and Lifestyle"
    )
    .when(
        category_value == "sports and travel",
        "Sports and Travel"
    )
    .when(
        category_value == "fashion accessories",
        "Fashion Accessories"
    )
    .otherwise("Unknown")
)
```

```python
# --------------------------------------------
# Standardize Product Name
# --------------------------------------------
df = df.withColumn(
    "Product_Name",
    when(
        lower(trim(col("Product_Name"))) == "hand bag",
        "Handbag"
    )
    .otherwise(trim(col("Product_Name")))
)
```

```python
# --------------------------------------------
# Standardize Payment Method
# --------------------------------------------
payment_value = lower(
    regexp_replace(
        trim(col("Payment_Method")),
        r"\s+", # Replaces multiple spaces with one space.
        ""
    )
)

df = df.withColumn(
    "Payment_Method",
    when(
        payment_value.isin(
            "cash",
            "cashpayment"
        ),
        "Cash"
    )
    .when(
        payment_value.isin(
            "creditcard",
            "card"
        ),
        "Credit Card"
    )
    .when(
        payment_value == "debitcard",
        "Debit Card"
    )
    .when(
        payment_value == "mobilewallet",
        "Mobile Wallet"
    )
    .otherwise("Unknown")
)
```

### *Converting the strings into numbers*

```python
# --------------------------------------------
# Convert Numeric Columns Safely
# --------------------------------------------

df = df.withColumn(
    "Sale_Row_ID",
    expr("try_cast(Sale_Row_ID as BIGINT)")
)

df = df.withColumn(
    "Unit_Price",
    expr("try_cast(Unit_Price as DOUBLE)")
)

df = df.withColumn(
    "Quantity",
    expr("try_cast(Quantity as INT)")
)

df = df.withColumn(
    "Discount_Percent",
    expr("try_cast(Discount_Percent as DOUBLE)")
)

df = df.withColumn(
    "Tax_Rate",
    expr("try_cast(Tax_Rate as DOUBLE)")
)

df = df.withColumn(
    "Cost_Per_Unit",
    expr("try_cast(Cost_Per_Unit as DOUBLE)")
)

df = df.withColumn(
    "Customer_Rating",
    expr("try_cast(Customer_Rating as DOUBLE)")
)
```

### *Fixing day date time Formates*

```python
# --------------------------------------------
# Convert Different Date Formats (Try several possible date formats and take the first one that works.)
# --------------------------------------------
parsed_date = coalesce(
    try_to_timestamp(
        col("Date"),
        lit("yyyy-MM-dd")
    ),
    try_to_timestamp(
        col("Date"),
        lit("dd/MM/yyyy")
    ),
    try_to_timestamp(
        col("Date"),
        lit("MM-dd-yyyy")
    ),
    try_to_timestamp(
        col("Date"),
        lit("yyyy/MM/dd")
    )
)

df = df.withColumn(
    "Date",
    to_date(parsed_date)
)
```

```python
# --------------------------------------------
# Convert Different Time Formats
# --------------------------------------------
df = df.withColumn(
    "Time_Timestamp",
    coalesce(      # Try several possible time formats and take the first one that works
        try_to_timestamp(
            col("Time"),
            lit("HH:mm")
        ),
        try_to_timestamp(
            col("Time"),
            lit("h:mm a")
        ),
        try_to_timestamp(
            col("Time"),
            lit("HH.mm")
        )
    )
)

# Store the clean time as HH:mm
df = df.withColumn(
    "Time",
    date_format(
        col("Time_Timestamp"),
        "HH:mm"
    )
)
```

```python
# --------------------------------------------
# Validate Unit Price
# --------------------------------------------
df = df.withColumn(
    "Unit_Price",
    when(
        col("Unit_Price").between(0.01, 10000), #above than this is an outlier 
        col("Unit_Price")
    ).otherwise(None)
)

# --------------------------------------------
# Validate Quantity
# --------------------------------------------
df = df.withColumn(
    "Quantity",
    when(
        col("Quantity").between(1, 100),
        col("Quantity")
    ).otherwise(None)
)

# --------------------------------------------
# Validate Discount
# --------------------------------------------
df = df.withColumn(
    "Discount_Percent",
    when(
        col("Discount_Percent").between(0, 100),
        col("Discount_Percent")
    ).otherwise(None)
)

# --------------------------------------------
# Validate Tax Rate
# --------------------------------------------
df = df.withColumn(
    "Tax_Rate",
    when(
        col("Tax_Rate").between(0, 30),
        col("Tax_Rate")
    ).otherwise(None)
)

# --------------------------------------------
# Validate Cost Per Unit
# --------------------------------------------
df = df.withColumn(
    "Cost_Per_Unit",
    when(
        (col("Cost_Per_Unit") > 0) &
        (col("Cost_Per_Unit") <= col("Unit_Price")),
        col("Cost_Per_Unit")
    ).otherwise(None)
)

# --------------------------------------------
# Validate Customer Rating
# --------------------------------------------
df = df.withColumn(
    "Customer_Rating",
    when(
        col("Customer_Rating").between(1, 10),
        col("Customer_Rating")
    ).otherwise(None)
)
```

### *Dealing with null values*


```python
# --------------------------------------------
# Fill Default Values
# --------------------------------------------
df = df.fillna({
    "Discount_Percent": 0.0,
    "Tax_Rate": 14.0,
    "Customer_Type": "Unknown",
    "Gender": "Unknown",
    "Product_Category": "Unknown",
    "Payment_Method": "Unknown"
})
```

```python
# --------------------------------------------
# Remove Rows Missing Critical Information
# --------------------------------------------
critical_columns = [
    "Invoice_ID",
    "Date",
    "Branch",
    "City",
    "Product_Name",
    "Unit_Price",
    "Quantity",
    "Cost_Per_Unit"
]

rows_before_missing_removal = df.count()

df = df.dropna(
    subset=critical_columns
)

rows_after_missing_removal = df.count()

print(
    "Rows Removed Due to Missing Critical Data:",
    rows_before_missing_removal -
    rows_after_missing_removal
)
```

```python
# --------------------------------------------
# Remove Business Duplicates
# --------------------------------------------
duplicate_columns = [                
    column_name
    for column_name in df.columns
    if column_name not in [
        "Sale_Row_ID",
        "Time_Timestamp"
    ]
]

rows_before_duplicates = df.count()

df = df.dropDuplicates(
    duplicate_columns
)

rows_after_duplicates = df.count()

print(
    "Duplicate Rows Removed:",
    rows_before_duplicates -
    rows_after_duplicates
)
```

```python
# --------------------------------------------
# Date Features
# --------------------------------------------
df = df.withColumn(
    "Year",
    year(col("Date"))
)

df = df.withColumn(
    "Quarter",
    quarter(col("Date"))
)

df = df.withColumn(
    "Month_Number",
    month(col("Date"))
)

df = df.withColumn(
    "Month_Name",
    date_format(col("Date"), "MMMM")
)

df = df.withColumn(
    "Month_Year",
    date_format(col("Date"), "yyyy-MM")
)

df = df.withColumn(
    "Day",
    dayofmonth(col("Date"))
)

df = df.withColumn(
    "Day_Name",
    date_format(col("Date"), "EEEE")
)

df = df.withColumn(
    "Week_Number",
    weekofyear(col("Date"))
)
```

```python
# --------------------------------------------
# Extract Hour
# --------------------------------------------
df = df.withColumn(
    "Hour",
    hour(col("Time_Timestamp"))
)

# --------------------------------------------
# Create Time Period
# --------------------------------------------
df = df.withColumn(
    "Time_Period",
    when(
        col("Hour").isNull(),
        "Unknown"
    )
    .when(
        col("Hour").between(6, 11),
        "Morning"
    )
    .when(
        col("Hour").between(12, 16),
        "Afternoon"
    )
    .when(
        col("Hour").between(17, 20),
        "Evening"
    )
    .otherwise("Night")
)
```

```python
# --------------------------------------------
# Subtotal
# --------------------------------------------
df = df.withColumn(
    "Subtotal",
    round(
        col("Unit_Price") *
        col("Quantity"),
        2
    )
)

# --------------------------------------------
# Discount Amount
# --------------------------------------------
df = df.withColumn(
    "Discount_Amount",
    round(
        col("Subtotal") *
        col("Discount_Percent") / 100,
        2
    )
)

# --------------------------------------------
# Net Sales
# --------------------------------------------
df = df.withColumn(
    "Net_Sales",
    round(
        col("Subtotal") -
        col("Discount_Amount"),
        2
    )
)

# --------------------------------------------
# Tax Amount
# --------------------------------------------
df = df.withColumn(
    "Tax_Amount",
    round(
        col("Net_Sales") *
        col("Tax_Rate") / 100,
        2
    )
)

# --------------------------------------------
# Final Total
# --------------------------------------------
df = df.withColumn(
    "Final_Total",
    round(
        col("Net_Sales") +
        col("Tax_Amount"),
        2
    )
)
```

```python
# --------------------------------------------
# Total Cost
# --------------------------------------------
df = df.withColumn(
    "Total_Cost",
    round(
        col("Cost_Per_Unit") *
        col("Quantity"),
        2
    )
)

# --------------------------------------------
# Profit
# --------------------------------------------
df = df.withColumn(
    "Profit",
    round(
        col("Net_Sales") -
        col("Total_Cost"),
        2
    )
)

# --------------------------------------------
# Profit Margin
# --------------------------------------------
df = df.withColumn(
    "Profit_Margin",
    when(
        col("Net_Sales") > 0,
        round(
            col("Profit") /
            col("Net_Sales") * 100,
            2
        )
    ).otherwise(0)
)
```

```python
# --------------------------------------------
# Rating Category
# --------------------------------------------
df = df.withColumn(
    "Rating_Category",
    when(
        col("Customer_Rating").isNull(),
        "Unknown"
    )
    .when(
        col("Customer_Rating") < 5,
        "Poor"
    )
    .when(
        col("Customer_Rating") < 7,
        "Average"
    )
    .when(
        col("Customer_Rating") < 8.5,
        "Good"
    )
    .otherwise("Excellent")
)
```

```python
# --------------------------------------------
# Quantity Category
# --------------------------------------------
df = df.withColumn(
    "Quantity_Category",
    when(
        col("Quantity") <= 2,
        "Small Purchase"
    )
    .when(
        col("Quantity") <= 5,
        "Medium Purchase"
    )
    .otherwise("Large Purchase")
)
```

```python
# --------------------------------------------
# Remove Temporary Column
# --------------------------------------------
df = df.drop("Time_Timestamp")
```

```python
# --------------------------------------------
# Select Final Columns
# --------------------------------------------
df_clean = df.select(
    "Sale_Row_ID",
    "Invoice_ID",
    "Date",
    "Time",

    "Year",
    "Quarter",
    "Month_Number",
    "Month_Name",
    "Month_Year",
    "Week_Number",
    "Day",
    "Day_Name",
    "Hour",
    "Time_Period",

    "Branch",
    "City",
    "Customer_Type",
    "Gender",

    "Product_Category",
    "Product_Name",

    "Unit_Price",
    "Quantity",
    "Quantity_Category",

    "Discount_Percent",
    "Discount_Amount",

    "Tax_Rate",
    "Tax_Amount",

    "Cost_Per_Unit",
    "Total_Cost",

    "Subtotal",
    "Net_Sales",
    "Final_Total",
    "Profit",
    "Profit_Margin",

    "Payment_Method",
    "Customer_Rating",
    "Rating_Category"
)
```

```python
# --------------------------------------------
# Row Count Comparison
# --------------------------------------------
clean_row_count = df_clean.count()

print("Raw Rows:", raw_row_count)
print("Clean Rows:", clean_row_count)
print("Rows Removed:", raw_row_count - clean_row_count)
```

```python
# --------------------------------------------
# Final Schema
# --------------------------------------------
df_clean.printSchema()
```

```python
# --------------------------------------------
# Final Missing Values Report
# --------------------------------------------
missing_values_after = df_clean.select([
    sum(
        when(
            col(column_name).isNull(),
            1
        ).otherwise(0)
    ).alias(column_name)
    for column_name in df_clean.columns
])

display(missing_values_after)
```
